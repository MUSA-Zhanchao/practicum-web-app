```python
import geopandas as gpd
import pandas as pd
import leafmap.foliumap as leafmap
```

```python
approach_gdf = gpd.read_file("../../data/output/approach_crossing_widths_baseline_filtered.gpkg")
approach_gdf.head()
```

```python
approach_gdf.columns.tolist()
```

```python
approach_wgs = approach_gdf.to_crs(epsg=4326)
approach_wgs.head()
```

```python
approach_wgs["crossing_width_ft"].describe()
```

```python
approach_wgs.nlargest(10, "crossing_width_ft")[["approach_id", "intersection_id", "crossing_width_ft", "valid_samples"]]
```

```python
m = leafmap.Map(center=[39.95, -75.19], zoom=15)
m
```

```python
m = leafmap.Map(center=[39.95, -75.19], zoom=15)

m.add_gdf(
    approach_wgs,
    layer_name="Crossing Width (raw)",
    info_mode="on_click"
)

m
```

```python
def width_class(ft):
    if ft < 20:
        return "<20 ft"
    elif ft < 30:
        return "20–30 ft"
    elif ft < 40:
        return "30–40 ft"
    elif ft < 50:
        return "40–50 ft"
    elif ft < 65:
        return "50–65 ft"
    else:
        return "65–80 ft"

approach_wgs["width_class"] = approach_wgs["crossing_width_ft"].apply(width_class)
approach_wgs["width_class"].value_counts()
```

```python
m = leafmap.Map(center=[39.95, -75.19], zoom=15)

m.add_gdf(
    approach_wgs,
    layer_name="Crossing Width Classes",
    info_mode="on_click"
)

m
```

```python
wide = approach_wgs[approach_wgs["crossing_width_ft"] >= 50].copy()

m = leafmap.Map(center=[39.95, -75.19], zoom=15)
m.add_gdf(
    wide,
    layer_name="Wide Crossings >= 50 ft",
    info_mode="on_click"
)
m
```
